```python
def get_data():
    sales = pd.read_csv(
        filepath_or_buffer="https://gist.githubusercontent.com/andron23/0675d643fe001b08a66271402a7b609b/raw/231deb6c35c43ca9089b197d94d7d48e093bfb06/sales_data.csv",
        parse_dates=["order_date"]
    )
    product = pd.read_csv(
        filepath_or_buffer="https://gist.githubusercontent.com/andron23/fc2b28de07d5017f951c559f51fcb286/raw/be9f8cd81b28f3c19d314ef908aa68e9f41ae975/products_data.csv",
    )
    return (sales, product)

```

```python
sal, prod = get_data()
```

```python
sal.head()
```

```python
sal.groupby("customer_id").agg({"total_price":"sum"}).reset_index().sort_values(by="total_price")
```

```python
inner = pd.merge(
    left=sales,
    right=prod,
    left_on="product_id",
    right_on="product_id",
    how="inner"
)

total_revenue = inner.total_price.sum()
tablet_revenue = inner[inner.category == "Планшеты"].total_price.sum()
(tablet_revenue / total_revenue).round(2)
```

```python
inner.groupby("region").agg({"total_price":"sum"}).reset_index()
regs = ("Восток", "Юг")
inner[inner.region.isin(regs) & (inner.category == "Смартфоны")].groupby("region").agg({"total_price":"mean"})
```
